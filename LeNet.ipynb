```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten,AveragePooling2D
from keras import Sequential
from keras.datasets import mnist
```

```python
(x_train,y_train),(x_test,y_test) = mnist.load_data()
```

```python
x_test.shape
```

```python
import numpy as np
from tensorflow.keras.utils import to_categorical

# Reshape to add a channel dimension (for grayscale images, this is 1)
x_train_reshaped = x_train.reshape(x_train.shape[0], 28, 28, 1)
x_test_reshaped = x_test.reshape(x_test.shape[0], 28, 28, 1)

# Pad images from 28x28 to 32x32
x_train_padded = np.pad(x_train_reshaped, ((0,0),(2,2),(2,2),(0,0)), 'constant')
x_test_padded = np.pad(x_test_reshaped, ((0,0),(2,2),(2,2),(0,0)), 'constant')

# Normalize pixel values to be between 0 and 1
x_train_normalized = x_train_padded.astype('float32') / 255.0
x_test_normalized = x_test_padded.astype('float32') / 255.0

# One-hot encode y_train and y_test
y_train_encoded = to_categorical(y_train, num_classes=10)
y_test_encoded = to_categorical(y_test, num_classes=10)


print("Shape of x_train_normalized:", x_train_normalized.shape)
print("Shape of x_test_normalized:", x_test_normalized.shape)
print("Shape of y_train_encoded:", y_train_encoded.shape)
print("Shape of y_test_encoded:", y_test_encoded.shape)
```

```python
model = Sequential()

model.add(Conv2D(6,kernel_size=(5,5),padding='valid',activation='tanh',input_shape=(32,32,1)))
model.add(AveragePooling2D(pool_size=(2,2),strides=2,padding='valid'))

model.add(Conv2D(16,kernel_size=(5,5),padding='valid',activation='tanh'))
model.add(AveragePooling2D(pool_size=(2,2),strides=2,padding='valid'))

model.add(Flatten())

model.add(Dense(120,activation='tanh'))
model.add(Dense(84,activation='tanh'))
model.add(Dense(10,activation='softmax'))

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model.summary()
```

```python
history = model.fit(x_train_normalized,
                    y_train_encoded,
                    epochs=30,
                    batch_size=32,
                    validation_data=(x_test_normalized,y_test_encoded))
```
